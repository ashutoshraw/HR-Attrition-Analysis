```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns',None)
pd.set_option('display.width',1000)
```

```python
df=pd.read_csv('hr_attrition.csv')
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
# Check columns with only ONE unique value (useless for analysis)
for col in df.columns:
    if df[col].nunique() == 1:
        print(f"{col} : {df[col].unique()}")
        
```

```python
# Check for duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.drop(columns=['EmployeeCount', 'StandardHours', 'Over18'], inplace=True)
print("Remaining columns:", df.shape[1])
```

```python
print(df['Attrition'].value_counts())
print("\nAttrition Rate:", round(df['Attrition'].value_counts(normalize=True)['Yes'] * 100, 2), "%")
```

```python
df.describe()
```

```python
education_map = {
    1: 'Below College',
    2: 'College',
    3: 'Bachelor',
    4: 'Master',
    5: 'Doctor'
}

df['EducationLabel'] = df['Education'].map(education_map)
print(df['EducationLabel'].value_counts())
```

```python
df['AttritionFlag'] = df['Attrition'].map({'Yes': 1, 'No': 0})
print(df['AttritionFlag'].value_counts())
```

```python
df.to_csv('hr_attrition_cleaned.csv', index=False)
print("Cleaned data saved successfully!")
print("Shape:", df.shape)
```
